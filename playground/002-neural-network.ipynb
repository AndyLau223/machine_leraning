```python
import numpy as np

# --- 1. SETUP DATA ---
# Input X: 2 features (e.g., [0,0], [0,1], [1,0], [1,1])
# We transpose it so each column is a training example: (2 features, 4 examples)
X = np.array([[0, 0, 1, 1],
              [0, 1, 0, 1]])

# Target Y: (1 output, 4 examples)
Y = np.array([[0, 1, 1, 0]])

# --- 2. INITIALIZE PARAMETERS ---
def init_params():
    np.random.seed(42) # For consistent results
    W1 = np.random.randn(3, 2) * 0.1  # 3 neurons in hidden layer
    b1 = np.zeros((3, 1))
    W2 = np.random.randn(1, 3) * 0.1  # 1 neuron in output layer
    b2 = np.zeros((1, 1))
    return W1, b1, W2, b2

# --- 3. MATH FUNCTIONS ---
def relu(Z):
    return np.maximum(0, Z)

def relu_derivative(Z):
    return Z > 0

def sigmoid(Z):
    return 1 / (1 + np.exp(-Z))

# --- 4. TRAINING LOOP ---
W1, b1, W2, b2 = init_params()
learning_rate = 0.1
m = X.shape[1] # Number of examples (4)

print("Starting Training...\n")

for i in range(10010):
    # FORWARD PROPAGATION
    Z1 = np.dot(W1, X) + b1        # (3,2) dot (2,4) = (3,4)
    A1 = relu(Z1)                  # (3,4)
    Z2 = np.dot(W2, A1) + b2        # (1,3) dot (3,4) = (1,4)
    A2 = sigmoid(Z2)               # (1,4) - Final Probabilities

    # CALCULATE LOSS (Binary Cross Entropy)
    loss = -np.mean(Y * np.log(A2 + 1e-8) + (1-Y) * np.log(1-A2 + 1e-8))

    # BACKWARD PROPAGATION (The "Gradient" calculation)
    dZ2 = A2 - Y                   # Error at output
    dW2 = (1/m) * np.dot(dZ2, A1.T)
    db2 = (1/m) * np.sum(dZ2, axis=1, keepdims=True)

    dZ1 = np.dot(W2.T, dZ2) * relu_derivative(Z1) # Error at hidden layer
    dW1 = (1/m) * np.dot(dZ1, X.T)
    db1 = (1/m) * np.sum(dZ1, axis=1, keepdims=True)

    # UPDATE PARAMETERS (Gradient Descent)
    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1
    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2

    if i % 200 == 0:
        print(f"Iteration {i}: Loss = {loss:.4f}")

print("\nFinal Predictions after training:")
print(np.round(A2, 2)) # Should be close to [0, 1, 1, 0]
```
