```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
```

```python
# 1. Create a mock dataset
rng = np.random.RandomState(42)
X = np.sort(5 * rng.rand(80,1), axis=0)
y = np.sin(X).ravel()
y[::5] += 3 * (0.5 - rng.rand(16)) # Add noise to every 5th data point

print(X.shape)
print(y.shape)
print(y)
```

```python
# 2. Fit Regression Models
regr_1 = DecisionTreeRegressor(max_depth=2)
regr_2 = DecisionTreeRegressor(max_depth=5)
regr_1.fit(X,y)
regr_2.fit(X,y)
```

```python
# 3. Predict
X_test = np.arange(0.0,5.0,0.01)[:, np.newaxis]
y_1 = regr_1.predict(X_test)
y_2 = regr_2.predict(X_test)
```

```python
# 4. Plot the results
plt.figure(figsize=(10,6))
plt.scatter(X,y, s=20, edgecolors="black", c="darkorange", label="data")
plt.plot(X_test, y_1, color="cornflowerblue", label="max_depth=2", linewidth=2)
plt.plot(X_test, y_2, color="yellowgreen", label="max_depth=5", linewidth=2)
plt.xlabel("data")
plt.ylabel("target")
plt.title("Decision Tree Regression: Capturing Non-Linear Patterns")
plt.legend()
plt.show()
```

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(15,10))
plot_tree(regr_1, filled=True, feature_names=["Input Value"], rounded=True, precision=2)
plt.title("Decision Logic for max_depth=2")
plt.show()
```

```python
plt.figure(figsize=(75,50))
plot_tree(regr_2, filled=True, feature_names=["Input Value"], rounded=True, precision=2)
plt.title("Decision Logic for max_depth=5")
plt.show()
```

## TensorFlow implementation of collaborative filtering

```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model
import matplotlib.pyplot as plt

# --- 1. FAST DATA LOADING ---
print("Loading data...")
url = "https://files.grouplens.org/datasets/movielens/ml-100k/u.data"
columns = ['user_id', 'item_id', 'rating', 'timestamp']

# We only take the first 20,000 rows to make sure it finishes quickly
df = pd.read_csv(url, sep='\t', names=columns, nrows=20000)

# --- 2. QUICK PREPROCESS ---
user_ids = df['user_id'].unique().tolist()
user2user_encoded = {x: i for i, x in enumerate(user_ids)}
movie_ids = df['item_id'].unique().tolist()
movie2movie_encoded = {x: i for i, x in enumerate(movie_ids)}

df['user'] = df['user_id'].map(user2user_encoded)
df['movie'] = df['item_id'].map(movie2movie_encoded)

num_users = len(user2user_encoded)
num_movies = len(movie2movie_encoded)

x = df[['user', 'movie']].values
y = df['rating'].values.astype("float32")

# --- 3. THE MODEL ---
user_input = layers.Input(shape=(1,))
movie_input = layers.Input(shape=(1,))

# Smaller embedding size (8 instead of 50) makes it much faster
user_emb = layers.Embedding(num_users, 8)(user_input)
movie_emb = layers.Embedding(num_movies, 8)(movie_input)

dot_product = layers.Dot(axes=2)([user_emb, movie_emb])
output = layers.Flatten()(dot_product)

model = Model(inputs=[user_input, movie_input], outputs=output)
model.compile(optimizer='adam', loss='mean_squared_error')

# --- 4. TRAIN AND LOG ---
print("Training starting (5 epochs)...")
# verbose=1 ensures you see the progress bar move
history = model.fit(
    x=[x[:, 0], x[:, 1]],
    y=y,
    batch_size=128,
    epochs=5,
    verbose=1 
)

# --- 5. PLOT THE LOG ---
plt.plot(history.history['loss'])
plt.title('Training Loss')
plt.ylabel('MSE')
plt.xlabel('Epoch')
plt.show()
```
