# Logistic Regression

## Sigmoid function

The sigmoid function (also called the logistic function) is a mathematical function that maps any real number into a value between 0 and 1.

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

- $z$ is the input (any real number).
- $e$ is Euler's number (4$\approx 2.718$).
- $\sigma(z)$ is the output (always between 0 and 1).6

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
z = np.linspace(-10,10,100)
y = sigmoid(z)
```

```python
plt.figure(figsize=(8, 4))
plt.plot(z, y, color='blue', label=r'$\sigma(z) = \frac{1}{1 + e^{-z}}$')
plt.title('The Sigmoid Function')
plt.xlabel('Input z')
plt.ylabel('Output $\sigma(z)$')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0.5, color='red', linestyle='--', label='Decision Threshold (0.5)')
plt.axvline(0, color='black', linewidth=0.8)
plt.legend()
plt.show()
```
