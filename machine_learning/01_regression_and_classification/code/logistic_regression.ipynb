```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z):
    return  1 / (1 + np.exp(-z))
    
```

```python
value = 0

print (f"sigmoid({value}) = {sigmoid(value)}")
print ("sigmoid([ -1, 0, 1, 2]) = " + str(sigmoid(np.array([-1, 0, 1, 2]))))
```

```python
def compute_cost(X, y, w, b):
    m, n = X.shape
    
    y_pred = sigmoid(np.dot(X,w) + b)
    
    total_cost = (1 / m) * np.sum((-y) * np.log(y_pred) - (1 - y) * np.log(1 - y_pred))
    return total_cost
```

```python
# UNQ_C3
# GRADED FUNCTION: compute_gradient
def compute_gradient(X, y, w, b, *argv): 
    m, n = X.shape
    dj_dw = np.zeros((n,))
    dj_db = 0.

    ### START CODE HERE ### 
    for i in range(m):
        # Calculate the prediction for the i-th example
        # Use dot product for the features of one row
        z_i = np.dot(X[i], w) + b
        f_wb_i = sigmoid(z_i)
        
        # Calculate the error for the i-th example
        err_i = f_wb_i - y[i]
        
        # Add to the bias gradient
        dj_db = dj_db + err_i
        
        # Add to the weight gradient for each feature j
        for j in range(n):
            dj_dw[j] = dj_dw[j] + err_i * X[i, j]
            
    # Divide by m to get the average
    dj_dw = dj_dw / m
    dj_db = dj_db / m
    ### END CODE HERE ###

    return dj_db,dj_dw 
```
